```python
import time
```

```python
from main import search_query
```

```python
queries=["How Clickhouse is better than Oracle?",
         "What is the vector search?",
         "How can I make my query faster?",
         "What databases support full-text search?",
         "How indexing can help me in database?",
         "What is the difference between a star schema and a snowflake schema?",
         "What is the difference between BTree and Bitmap indexes?",
         "What is the columnar storage in databases?",
         "What is the difference between OLAP and OLTP?",
         "How to partition table in database?"]
```

```python
durations=[]
```

```python
for query in queries:
    start = time.perf_counter()
    search_query(query)
    end = time.perf_counter()
    duration=end-start
    durations.append(duration)
```

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df=pd.DataFrame(durations)
```

```python
df
```

```python
df.plot(kind='bar',title='Длительность запроса',legend=False)
```

```python
df.plot(kind='kde',title='Длительность запроса',legend=False)
```
